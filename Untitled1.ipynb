```python
import numpy as np

# 1. Student names
students = np.array(["sravya", "sree", "madhav", "satya", "hima"])

# 2. Student marks: Maths, Science, English
marks = np.array([
    [85, 90, 78],
    [60, 75, 80],
    [95, 92, 88],
    [45, 55, 50],
    [75, 80, 85]
])

subjects = np.array(["Maths", "Science", "English"])

print("===== STUDENT PERFORMANCE ANALYSIS =====")

# 3. Display student marks
print("\nStudent Marks:")
print(marks)

# 4. Subject-wise average
subject_avg = np.mean(marks, axis=0)

print("\nSubject-wise Average:")
for i in range(len(subjects)):
    print(subjects[i], ":", subject_avg[i])

# 5. Student-wise average
student_avg = np.mean(marks, axis=1)

print("\nStudent-wise Average:")
for i in range(len(students)):
    print(students[i], ":", round(student_avg[i], 2))

# 6. Highest and lowest marks
print("\nHighest Mark:", np.max(marks))
print("Lowest Mark:", np.min(marks))

# 7. Topper identification
index = np.unravel_index(np.argmax(marks), marks.shape)

print("\nHighest Individual Score:")
print("Student:", students[index[0]])
print("Subject:", subjects[index[1]])
print("Marks:", marks[index])

# 8. Subject-wise highest marks
print("\nSubject-wise Highest Marks:")
for i in range(len(subjects)):
    print(subjects[i], ":", np.max(marks[:, i]))

# 9. Subject-wise lowest marks
print("\nSubject-wise Lowest Marks:")
for i in range(len(subjects)):
    print(subjects[i], ":", np.min(marks[:, i]))

# 10. Total marks per student
total_marks = np.sum(marks, axis=1)

print("\nTotal Marks:")
for i in range(len(students)):
    print(students[i], ":", total_marks[i], "/ 300")

# 11. Standard deviation per subject
std_marks = np.std(marks, axis=0)

print("\nSubject-wise Standard Deviation:")
for i in range(len(subjects)):
    print(subjects[i], ":", round(std_marks[i], 2))

# 12. Students above overall average
overall_avg = np.mean(marks)

print("\nOverall Average:", round(overall_avg, 2))
print("Students Above Overall Average:")

for i in range(len(students)):
    if student_avg[i] > overall_avg:
        print(students[i])

# 13. Pass/Fail analysis
# Assumption: Each subject lo 50 marks vaste pass
pass_status = np.all(marks >= 50, axis=1)

print("\nPass/Fail Results:")
for i in range(len(students)):
    if pass_status[i]:
        print(students[i], ": Pass")
    else:
        print(students[i], ": Fail")

# 14. Overall topper based on total marks
topper_index = np.argmax(total_marks)

print("\nOverall Topper:", students[topper_index])
print("Total Marks:", total_marks[topper_index], "/ 300")

# 15. Overall performance percentage
percentage = (total_marks / 300) * 100

print("\nStudent Percentages:")
for i in range(len(students)):
    print(students[i], ":", round(percentage[i], 2), "%")

```
